```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, MiniBatchKMeans
from sklearn.metrics import silhouette_score
```

# kmean

* KMean

```python
xb, _ = make_blobs(n_samples=20000, n_features=10, centers= 8, cluster_std=1.1, random_state=42)

plt.scatter(xb[:,0], xb[:,1])
```

```python
model = KMeans(n_clusters=8, init='k-means++', n_init='auto', max_iter=300)
model.fit(xb)
cluster = model.predict(xb)
plt.scatter(xb[:,0], xb[:,1], c=cluster, label=cluster)
plt.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1], c="r")
plt.legend()
```

* MinibatchKMean

```python
model = MiniBatchKMeans(n_clusters=8, init='k-means++', max_iter=100, batch_size=1024)
model.fit(xb)
cluster = model.predict(xb)
plt.scatter(xb[:,0], xb[:,1], c=cluster, label=cluster)
plt.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1], c="r")
plt.legend()
```

# DBSCAN

```python
from sklearn.cluster import DBSCAN
```

```python
model = DBSCAN(eps=5, min_samples=5, metric='euclidean', algorithm='auto', leaf_size=30)

cluster = model.fit_predict(xb)
plt.scatter(xb[:,0], xb[:,1], c=cluster)
#plt.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1], c="r")
```

# Mean-Shift

```python
from sklearn.cluster import MeanShift
```

```python
model = MeanShift(min_bin_freq=1, cluster_all=True)
cluster = model.fit_predict(xb)
plt.scatter(xb[:,0], xb[:,1], c=cluster)
```

# Spectral clustring

```python
from sklearn.cluster import SpectralClustering
```

```python
model = SpectralClustering(n_clusters=8, 
                           n_init=10, gamma=1, affinity='rbf', 
                           n_neighbors=10, assign_labels='kmeans')
cluster = model.fit_predict(xb)
plt.scatter(xb[:,0], xb[:,1], c=cluster)
```

# Hierarchical clustering

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
model = AgglomerativeClustering(n_clusters=8, metric='euclidean', linkage='ward')
cluster = model.fit_predict(xb)
plt.scatter(xb[:,0], xb[:,1], c=cluster)
```
